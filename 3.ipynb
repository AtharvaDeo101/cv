```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
img1 = cv2.imread("C:/Users/deoat/Pictures/Screenshots/Screenshot 2026-07-17 125041.png")
img2= cv2.imread("C:/Users/deoat/Pictures/Screenshots/Screenshot 2026-07-17 125054.png")
```

```python
sift = cv2.SIFT_create()

kp1,des1 = sift.detectAndCompute(img1,None)
kp2, des2 = sift.detectAndCompute(img2,None)# detect features 
```

```python
bf = cv2.BFMatcher()
matches = bf.knnMatch(des1,des2,k=2)# match features
```

```python
good = [m for m, n in matches if m.distance < 0.75 * n.distance]
```

```python
#calculate Homography
src = np.float32([kp1[m.queryIdx].pt for m in good]).reshape(-1,1,2)
dst = np.float32([kp2[m.trainIdx].pt for m in good]).reshape(-1,1,2)
```

```python
H, mask = cv2.findHomography(src,dst,cv2.RANSAC,5.0)
print(H)#- It can be used to align or transform one image to match the other.
```
