```python
import cv2
import numpy as np
```

```python
img = cv2.imread("C:/Users/deoat/Downloads/Ebony-Wood-chess-board_2000x.webp")
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
```

```python
ret, corners = cv2.findChessboardCorners(gray, (9, 6), None)
```

```python
if ret:
    print("Chessboard detected!")

    # Camera calibration
    obj = np.zeros((54, 3), np.float32)
    obj[:, :2] = np.mgrid[0:9, 0:6].T.reshape(-1, 2)

    ret, matrix, dist, rvecs, tvecs = cv2.calibrateCamera(
        [obj], [corners], gray.shape[::-1], None, None
    )

    print("Camera Matrix:\n", matrix)
    print("Distortion:\n", dist)
else:
    print("Chessboard not detected!")
```

```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread("C:/Users/deoat/Downloads/istockphoto-1302940862-612x612.jpg")
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Detect chessboard corners
ret, corners = cv2.findChessboardCornersSB(gray, (9, 6))

if ret:
    print("Chessboard detected!")
    cv2.drawChessboardCorners(img, (9, 6), corners, ret)

    plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    plt.show()
else:
    print("Chessboard not detected!")
```
