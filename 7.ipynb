```python
import cv2
import matplotlib.pyplot as plt 
```

```python
img = cv2.imread("C:/Users/deoat/Downloads/gettyimages-172992650-612x612.jpg")
gray = cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
```

```python
edges = cv2.Canny(gray,100,200)
```

```python
lines = cv2.HoughLinesP(edge, 1, 3.14/180, 100, minLineLength=50, maxLineGap = 10)
```

```python
line_img = img.copy()
```

```python
if lines is not None:
    for line in lines:
        x1, y1 , x2, y2 = line[0]
        cv2.line(line_img,(x1,y1),(x2,y2),(0,))
```

```python
# 3. Corner Detection
corners = cv2.goodFeaturesToTrack(gray, 100, 0.01, 10)

corner_img = img.copy()
```

```python
if corners is not None:
    for c in corners:
        x, y = c.ravel().astype(int)
        cv2.circle(corner_img, (x,y), 5, (0,0,255), -1)
        
```

```python
plt.imshow(edges, cmap="gray")
plt.title("Edge Detection")
plt.show()

plt.imshow(cv2.cvtColor(line_img, cv2.COLOR_BGR2RGB))
plt.title("Line Detection")
plt.show()

plt.imshow(cv2.cvtColor(corner_img, cv2.COLOR_BGR2RGB))
plt.title("Corner Detection")
plt.show()
```
