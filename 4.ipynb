```python
import cv2
import matplotlib.pyplot as plt
import numpy as np 

img= cv2.imread("C:/Users/deoat/Downloads/Ck6vt.jpg")
img = cv2.cvtColor(img,cv2.COLOR_BGR2RGB)

h,w = img.shape[:2]
src = np.float32([
    [100,100],
    [w-100,100],
    [100,h-100],
    [w-100,h-100]
])

des = np.float32([
    [0,0],
    [300,0],
    [0,300],
    [300,300]
])


matrix = cv2.getPerspectiveTransform(src,des)

result = cv2.warpPerspective(img,matrix,(300,300))
plt.imshow(result)
```
