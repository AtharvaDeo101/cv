```python
import cv2 
import numpy as np 
```

```python
img1 = cv2.imread("C:/Users/deoat/Pictures/Screenshots/Screenshot 2026-07-17 125041.png")
img2= cv2.imread("C:/Users/deoat/Pictures/Screenshots/Screenshot 2026-07-17 125054.png")
```

```python
sift = cv2.SIFT_create()
kp1, des1 = sift.detectAndCompute(img1,None)
kp2,des2 = sift.detectAndCompute(img2,None)
```

```python
bf = cv2.BFMatcher()
watches = bf.knnMatch(des1, des2, k = 2 )
```

```python
good = [m for m, n in watches if m.distance < 0.75 * n.distance]
```

```python
pts1 = np.float32([kp1[m.queryIdx].pt for m in good])
pts2 = np.float32([kp2[m.queryIdx].pt for m in good])
```

```python
F , mask = cv2.findFundamentalMat(pts1,pts2,cv2.FM_RANSAC)

print(F)
```
